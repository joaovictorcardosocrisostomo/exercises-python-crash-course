```python
# Experimentando o matplotlib com gráfico linear simples
import matplotlib.pyplot as plt

input_values = [1, 2, 3, 4, 5]
squares = [1, 4, 9, 16, 25]
plt.plot(input_values, squares, linewidth=5)

# Define o título e nomeia os eixos
plt.title("Square Numbers Graphic", fontsize=24)
plt.xlabel("Value", fontsize=14)
plt.ylabel("Square of Value", fontsize=14)

# Define o tamanho dos rótulos das marcações
plt.tick_params(axis='both', labelsize=14)

plt.show()
```

```python
# Calculando dados automaticamente
x_values = list(range(1, 1001))
y_values = [x**2 for x in x_values]

plt.scatter(x_values, y_values, c='red', edgecolors='none', s=40) # o parâmetro edgecolor='none' remove o contorno dos pontos
# o parâmetro c='' define cores personalizadas pros pontos, permitindo formato RGB também (x, y, z)

plt.title("Dados Automáticos", fontsize=20)
plt.xlabel("Valor", fontsize=14)
plt.ylabel("Valor ao Quadrado", fontsize=14)

plt.tick_params(axis="both", which='major', labelsize=14)

plt.axis([0, 1100, 0, 1100000])

plt.show()
```

```python
# Aplicando Colormaps
""" Também é possível fazer o uso de colormaps. 
Um colormap é uma série de cores em um gradiente que varia de uma cor inicial até uma cor final. """

# Calculando dados automaticamente
x_values = list(range(1, 1001))
y_values = [x**2 for x in x_values]

plt.scatter(x_values, y_values, c=y_values, cmap=plt.cm.BrBG, edgecolors='none', s=40) # o parâmetro edgecolor='none' remove o contorno dos pontos
# o parâmetro c='' define cores personalizadas pros pontos, permitindo formato RGB também (x, y, z)

plt.title("Dados Automáticos", fontsize=20)
plt.xlabel("Valor", fontsize=14)
plt.ylabel("Valor ao Quadrado", fontsize=14)

plt.tick_params(axis="both", which='major', labelsize=14)

plt.axis([0, 1100, 0, 1100000])

# O Matplotlib funciona como um pintor desenhando em uma tela em branco.
# No Jupyter, quando a célula termina de rodar (ou quando você usa o comando plt.show()), 
# o Matplotlib "entrega" o quadro pronto para você ver e limpa a tela imediatamente,
# pegando uma nova tela em branco para o próximo desenho.
# O SAVEFIG VEM AQUI, COM A TELA AINDA PINTADA

plt.savefig('squares_plot.png', bbox_inches='tight') 

# O SHOW (SE FOR USAR) VEM POR ÚLTIMO, POIS ELE LIMPA A TELA
plt.show()
```

```python
# Implementando um random walk
# vamos gerar dados aleatórios para analisar em uma função RandomWalk()

from random import choice

class RandomWalk():
    """ Uma classe que gera passeios aleatórios. """
    def __init__(self, num_points=5000):
        self.num_points = num_points
        
        # o passeio sempre começa em (0,0)
        self.x_values = [0]
        self.y_values = [0]
    
    def fill_walk(self):
        """ Calcula os pontos do passeio. """
        # Continua os passos até chegar no número desejado de passos
        while len(self.x_values) < self.num_points:
            # Decide a direção e distância a ser percorrida
            x_direction = choice([1, -1])
            x_distance = choice([0, 1, 2, 3, 4])
            x_step = x_direction * x_distance
            
            y_direction = choice([1, -1])
            y_distance = choice([0, 1, 2, 3, 4])
            y_step = y_direction * y_distance
            
            if x_step == 0 and y_step == 0:
                continue
            
            next_x = self.x_values[-1] + x_step
            next_y = self.y_values[-1] + y_step
            
            self.x_values.append(next_x)
            self.y_values.append(next_y)

while True:
    rw = RandomWalk()
    rw.fill_walk()
    
    plt.scatter(rw.x_values, rw.y_values, edgecolors='none', s=15)
    plt.title("Random Walk", fontsize=20)
    plt.xlabel("X Value", fontsize=14)
    plt.ylabel("Y Value", fontsize=14)
    plt.show()
    
    keep_running = input("Make another walk? (y/n): ")
    if keep_running == 'n':
        break
```
